```python
import pandas as pd                          
import matplotlib.pyplot as plt 
import numpy as np
```

```python
%matplotlib inline                           
%config InlineBackend.figure_format='retina' 
plt.rcParams['figure.figsize'] = 12,6  
```

1. Read the CO2 datasets

```python
ees_climate_dir = 'C:/Users/TU_Climate24_Student/Desktop/ees_climate/'  #### This is the home directory we are always working out of

CO2_MaunaLoa = pd.read_csv(ees_climate_dir+'unit2_grisales/data/greenhouse_gases/CO2_MaunaLoa_TimeSeries.csv',parse_dates=['Date'])
CO2_NOAA = pd.read_csv(ees_climate_dir+'unit2_grisales/data/greenhouse_gases/CO2_NOAA_TimeSeries.csv',parse_dates=['Date'])
CO2_WDCGG = pd.read_csv(ees_climate_dir+'unit2_grisales/data/greenhouse_gases/CO2_WDCGG_TimeSeries.csv',parse_dates=['Date'])
```

2. Create three seperate plots of time series of CO2 concentration in ppm.

```python
fig, ax = plt.subplots(figsize=(5,5)) #Create a figure with a 5x5 size and plot it to this axis

ax.plot(CO2_MaunaLoa['Date'], # x axis
         CO2_MaunaLoa['Mauna Loa CO2 (ppm)'],# y axis
         linewidth=1, 
         label='Mauna Loa', #label to the plot
         color='black')

## Set properties of title, ylabels and xlabels
ax.set_title('Measured CO2 from Mauna Loa',
             fontsize=12, fontweight='bold')
ax.set_ylabel('CO2 concentration [ppm]',
             fontsize=12, fontweight='bold')
ax.set_xlabel('Year',
             fontsize=12, fontweight='bold')

## plot a legend on the ax
plt.legend() #Put a legend
```

```python
fig, ax = plt.subplots(figsize=(5,5)) #Create a figure with a 5x5 size and plot it to this axis

ax.plot(CO2_NOAA['Date'], # x axis
                 CO2_NOAA['NOAA CO2 (ppm)'],# y axis
         linewidth=1, 
         label='NOAA', #label to the plot
         color='black')

## Set properties of title, ylabels and xlabels
ax.set_title('Measured CO2 from NOAA',
             fontsize=12, fontweight='bold')
ax.set_ylabel('CO2 concentration [ppm]',
             fontsize=12, fontweight='bold')
ax.set_xlabel('Year',
             fontsize=12, fontweight='bold')

## plot a legend on the ax
plt.legend() #Put a legend
```

```python
fig, ax = plt.subplots(figsize=(5,5)) #Create a figure with a 5x5 size and plot it to this axis

ax.plot(CO2_WDCGG['Date'], # x axis
        CO2_WDCGG['WDCGG CO2 (ppm)'],# y axis
         linewidth=1, 
         label='WDCGG', #label to the plot
         color='black')

## Set properties of title, ylabels and xlabels
ax.set_title('Measured CO2 from WDCGG',
             fontsize=12, fontweight='bold')
ax.set_ylabel('CO2 concentration [ppm]',
             fontsize=12, fontweight='bold')
ax.set_xlabel('Year',
             fontsize=12, fontweight='bold')

## plot a legend on the ax
plt.legend() #Put a legend
```

3. Create a single plot with all three datasets on the plot, make each time series a seperate color and with a legend on the plot (hint .... you can use the fig, ax = plt.subplots(figsize=(5,5)) method for this ... one plot will plot on top of the other).

```python
fig, ax = plt.subplots(figsize=(5,5)) #Create a figure with a 5x5 size and plot it to this axis

ax.plot(CO2_MaunaLoa['Date'], # x axis
         CO2_MaunaLoa['Mauna Loa CO2 (ppm)'],# y axis
         linewidth=1, 
         label='Mauna Loa', #label to the plot
         color='black')
ax.plot(CO2_NOAA['Date'], # x axis
                 CO2_NOAA['NOAA CO2 (ppm)'],# y axis
         linewidth=2, 
         label='NOAA', #label to the plot
         color='indianred')
ax.plot(CO2_WDCGG['Date'], # x axis
        CO2_WDCGG['WDCGG CO2 (ppm)'],# y axis
         linewidth=2, 
         label='WDCGG', #label to the plot
         color='olive')

## Set properties of title, ylabels and xlabels
ax.set_title('Measured CO2',
             fontsize=12, fontweight='bold')
ax.set_ylabel('CO2 concentration [ppm]',
             fontsize=12, fontweight='bold')
ax.set_xlabel('Year',
             fontsize=12, fontweight='bold')

## plot a legend on the ax
plt.legend() #Put a legend
```

Figure 1. Monthly measured CO2 concentrations from three different datasets: Mauna Loa Observatory in Hawaii (black line), NOAA's Global Monitoring Laboratory (pink line) and World Data Centre for Greenhouse Gases (WDCGG) operated by the Japan Meteorological Agency
